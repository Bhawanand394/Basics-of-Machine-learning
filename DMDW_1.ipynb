```python
original_data = [1,2,3,2.7,5,5.2,7,8,7.9,10]
data = original_data.copy()
for i in range(len(data)-3 + 1):
  m = np.array(data[i:i+3]).mean()
  data[i+1] = m
```

```python
data
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 6))
plt.plot(original_data, marker='o', linestyle='--', color='red', label='Original Data')
plt.plot(data, marker='o', linestyle='-', color='blue', label='Smoothed Data')
plt.title('Original vs. Smoothed Data')
plt.xlabel('Index')
plt.ylabel('Value')
plt.grid(True)
plt.legend()
plt.show()
```
